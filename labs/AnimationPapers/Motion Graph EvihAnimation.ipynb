# Motion Graph EvihAnimation

This notebook rebuilds the Motion Graph case through the EvihAnimation/AI4AnimationPy BVH loader, keeps the original graph algorithm constants, and exports a Raylib-viewable playback artifact.

```python
import os
from pathlib import Path

ANIMATIONTECH_NOTEBOOK_CWD = Path(r".").resolve()
os.chdir(ANIMATIONTECH_NOTEBOOK_CWD)
print(ANIMATIONTECH_NOTEBOOK_CWD)
```

```python
import json
import sys

sys.path.insert(0, str(Path.cwd()))

import matplotlib.pyplot as plt
from matplotlib import colors as plt_colors
import numpy as np

from evih_motion_graph.core import GENERATED_ARTIFACT, run_pipeline, save_generated
```

## Run Equivalent Motion Graph Pipeline

```python
result = run_pipeline(strict_baseline=True)
artifact_path = save_generated(result, Path(GENERATED_ARTIFACT))
print(json.dumps(result.metrics, indent=2))
print(f"generated: {artifact_path}")
```

## Distance Matrix And Local Minima

```python
plt.figure(figsize=(7, 7))
plt.imshow(np.clip(result.distances, 1000, 15000), cmap=plt.cm.gray, interpolation="nearest")
plt.imshow(result.local_minima, cmap=plt_colors.ListedColormap(["none", "yellow"]), vmin=0, vmax=1, interpolation="nearest")
plt.title("EvihAnimation Motion Graph transition minima")
plt.show()
np.unique(result.local_minima, return_counts=True)
```
